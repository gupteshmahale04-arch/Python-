
# Q3. Multilevel Inheritance — Grandfather, Father, Son

## Problem Statement
Create:
- Grandfather class
- Father class inherited from Grandfather
- Son class inherited from Father

Store:
- grandfather_name
- father_name
- son_name

Display all details.

## Sample Output
```python
Grandfather: Mohan
Father: Rajesh
Son: Aman
```


````python
# Write your code here
class Grandfather:
    def __init__(self,G_Name):
        self.G_Name = G_Name
class Father (Grandfather):
    def __init__(self, G_Name,F_name):
        super().__init__(G_Name)
        self.F_name=F_name
class Son(Father):
    def __init__(self, G_Name, F_name,N_son):
        super().__init__(G_Name, F_name)
        self.N_Son=N_son

    def show(self):
        print(f"Grandfather :{self.G_Name}")
        print(f"Father :{self.F_name}")
        print(f"Son  : {self.N_Son}")
obj=Son("Maruti","Ambadas","Guptesh")
obj.show()
````


# Q7. Hierarchical Inheritance — Vehicle, Car, Bike

## Problem Statement
Create parent class `Vehicle`.

Create two child classes:
- Car
- Bike

Store:
- company_name
- model_name

Display details using both child classes.

## Sample Output
```python
Car Company: BMW
Car Model: X5

Bike Company: Yamaha
Bike Model: R15
```


````python
# Write your code here
class Vehicle:
    def __init__(self):
        print("This is class Vehicle")
class Bike(Vehicle):
    def __init__(self,Company,Model):
        super().__init__()
        self.Company=Company
        self.Model=Model
    def show (self):
        print(f"Bike Company :{self.Company}")
        print(f"Bike Model : {self.Model}")
class Car(Vehicle):
    def __init__(self,company,model):
        super().__init__()    
        self.company=company
        self.model=model
    def show (self):
        print(f"Car Company :{self.company}")
        print(f"Car Model : {self.model}")

obj_Bike=Bike("BMW",'X5')
obj_Bike.show()
obj_Car=Car("Yamaha","R15")
obj_Car.show()
        
````
